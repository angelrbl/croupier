### EDA of basic strategy | Croupier

```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv('../../../data/simulation_results.csv')

df['result'] = df['result'].astype(str).str.replace('Result.', '')
df['is_win'] = (df['result'] == 'win').astype(int)

df.head()
```

```python
win_rates = df['result'].value_counts(normalize=True)

fig1 = px.bar(
    x=win_rates.index, 
    y=win_rates.values,
    title="Results Distribution (Win/Loss/Draw)",
    labels={'x': 'Result', 'y': 'Percentage (%)'},
    color=win_rates.index,
    color_discrete_map={'win': '#2ecc71', 'loss': '#e74c3c', 'draw': '#95a5a6'}
)

fig1.show()
```

```python
bust_rate = df['player_bust'].value_counts(normalize=True)

fig2 = px.pie(
    names=["No Bust", "Bust"],
    values=bust_rate.values,
    title="Bust Distribution",
    color=['no_bust', 'bust'],
    color_discrete_map={'bust': '#e74c3c', 'no_bust': '#2ecc71'}
)

fig2.show()
```

```python
heatmap_data = df.pivot_table(
    index='player_initial_score',
    columns='dealer_upcard_value',
    values='is_win',
    aggfunc='mean'
)

px.imshow(
    heatmap_data,
    labels=dict(x="Dealer Upcard", y="Player Hand", color="Win Rate"),
    color_continuous_scale='RdYlGn',
    origin='lower'
)
```

---

#### We can clearly see that the problem with this strategy comes when the player has a relatively high hand (12-16) and the dealer has a low card, as he should be standing and hoping that the dealer busts.
